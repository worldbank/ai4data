# Frozen study samples

The original implementation sampled without a recorded random seed. The exact realized development and held-out samples are therefore the authoritative reproducibility artifacts. This notebook verifies those manifests.


```python
import pandas as pd

from schema_development.paths import ROOT
```

```python
development = pd.read_csv(ROOT / "data/manifests/development_snapshots.csv")
heldout = pd.read_csv(ROOT / "data/manifests/heldout_snapshots.csv")

assert len(development) == 210
assert len(heldout) == 202
assert set(development.snapshot_filename).isdisjoint(heldout.snapshot_filename)

display(development.groupby(["source", "artifact_type"]).size())
display(heldout.groupby(["source", "artifact_type"]).size())
```
